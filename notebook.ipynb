## Limpieza preliminar: Enero de 2014

```python
from sunpy.net import Fido, attrs as a

tstart = "2014/01/01 00:00:00"
tend   = "2014/01/31 23:59:59"

res = Fido.search(
    a.Time(tstart, tend),
    a.hek.EventType("FL"),
)
```

```python
fl = res["hek"]
print(len(fl))
print(fl.colnames)
```

```python
cols = ["event_starttime", "event_peaktime", "event_endtime", "fl_goescls", "fl_peakflux", "ar_noaanum", "obs_observatory", "frm_name"]
cols = [c for c in cols if c in fl.colnames]

df = fl[cols].to_pandas()
```

```python
print(df.head(), "\n")
print(df.isna().sum(), "\n")
print(df.dtypes, "\n")
print(df.obs_observatory.value_counts(), "\n")
print(df.frm_name.value_counts())
```

```python
g = (g.sort_values("ar_noaanum", ascending=False)
     .drop_duplicates(subset=["event_peaktime", "fl_goescls"], keep="first")
     .sort_values("event_peaktime")
     .reset_index(drop=True))

print(len(g), g["ar_noaanum"].isna().sum())
```

```python
print(g[g["ar_noaanum"] == 0]["fl_goescls"].str[0].value_counts())
```

```python
MULT = {"A": 1e-8, "B": 1e-7, "C": 1e-6, "M": 1e-5, "X": 1e-4}
COLS = ["event_starttime", "event_peaktime", "event_endtime", "fl_goescls", "fl_peakflux", "ar_noaanum", "obs_observatory", "frm_name", "hgs_x", "hgs_y"]

def get_goes_fl(t0, t1):
    # Filtro por GOES, cálculo de flujo en W/m2, limpieza de ar_noaanum = 0 y eliminación de duplicados
    fl = Fido.search(
        a.Time(t0, t1),
        a.hek.EventType("FL"))["hek"]
    cols = [c for c in COLS if c in fl.colnames]
    df = fl[cols].to_pandas()

    g = df[df["obs_observatory"] == "GOES"].copy()
    g["flux_wm2"] = g["fl_goescls"].str[0].map(MULT) * g["fl_goescls"].str[1:].astype(float)

    g["ar_noaanum"] = g["ar_noaanum"].replace(0, pd.NA)
    
    g = (g.sort_values("ar_noaanum", ascending=False)
         .drop_duplicates(subset=["event_peaktime", "fl_goescls"], keep="first")
         .sort_values("event_peaktime")
         .reset_index(drop=True))

    return g 

g = get_goes_fl(tstart, tend)
```

```python
print(len(g), g["ar_noaanum"].isna().sum())
```

* HEK devolvió 2327 eventos de tipo fulguración (flare) durante el mes de enero de 2014.
* 2156 de estas provienen del SDO (Solar Dynamics Observatory, con detección automática y sin clase GOES), mientras que 171 se atribuyen a GOES.
* Tras quitar 5 duplicados entre programas del SWPC quedaron 166 fulguraciones GOES.
* El flujo GOES es aproximado, puesto que se calculó con base en las clases (con dos cifras significativas Cx.x). Además, 24 fulguraciones (14%) se encontraron sin región activa registrada. GOES corresponde a un flujo de rayos X blandos, no de rayos X duros.
* No se encontró ningún evento RHESSI en HEK en 2014.

```python
meses = pd.date_range(tstart, "2014/12/01 00:00:00", freq="MS")
partes = []

for s in meses:
    e = s + pd.offsets.MonthEnd(0)
    try:
        partes.append(get_goes_fl(s.strftime("%Y/%m/%d"),
                                  e.strftime("%Y/%m/%d") + " 23:59:59"))
        print(s.strftime("%Y/%m"), len(partes[-1]))
    except Exception as err:
        print(s.strftime("%Y/%m"), "falló:", err)

goes2014 = pd.concat(partes, ignore_index=True)
# goes2014.to_csv("raw_data.csv", index=False)
print(len(goes2014))
```
